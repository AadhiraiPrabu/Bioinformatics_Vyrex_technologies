```python
 dna = 'ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA'
```

```python
count_A = dna.count('A')
count_T = dna.count('T')
count_G = dna.count('G')
count_C = dna.count('C')
```

```python
print(f'A: {count_A}')
print(f'T: {count_T}')
print(f'G: {count_G}')
print(f'C: {count_C}')
```

```python
gc = (count_G + count_C) / len(dna) * 100
print(f'GC Content: {gc:.2f}%')
```

```python
counts = {base: dna.count(base) for base in 'ATGC'}
print(counts)
```

```python
import matplotlib.pyplot as plt
```

```python
bases = list(counts.keys())
values = list(counts.values())
```

```python
plt.bar(bases, values)
plt.xlabel('Nucleotide')
plt.ylabel('Count')
plt.title('DNA Nucleotide Composition')
plt.show()
```
