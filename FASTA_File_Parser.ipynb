```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
```

```python
records = list(SeqIO.parse('insulin_human.fasta', 'fasta'))

print(f"Number of sequences: {len(records)}")
```

```python
for r in records:
    print(f"ID: {r.id}")
    print(f"Length: {len(r.seq)}")
    print(f"GC: {gc_fraction(r.seq) * 100:.2f}%")
    print(f"Sequence: {str(r.seq[:50])}...")
    print("-" * 50)
```

```python
import pandas as pd

summary = []

for r in records:
    summary.append({
        'ID': r.id,
        'Length': len(r.seq),
        'GC (%)': gc_fraction(r.seq) * 100
    })

df = pd.DataFrame(summary)

df
```
