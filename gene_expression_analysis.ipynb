```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('input.csv', index_col=0)

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)
```

```python
df = df.apply(pd.to_numeric, errors='coerce')
```

```python
df['mean_expression'] = df.mean(axis=1)

df.head()
```

```python
df.mean(axis=1)
```

```python
top10 = df['mean_expression'].sort_values(ascending=False).head(10)

print(top10)
```

```python
top10.plot(kind='barh', figsize=(8, 6))

plt.xlabel('Mean Expression')
plt.ylabel('Gene')
plt.title('Top 10 Expressed Genes')

plt.tight_layout()

plt.savefig('top_genes.png', dpi=300)

plt.show()
```

```python
top_gene = top10.index[0]
top_value = top10.iloc[0]

print(f"Highest expressed gene: {top_gene}")
print(f"Mean expression: {top_value:.2f}")
```
